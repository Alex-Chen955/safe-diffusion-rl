```python
import numpy as np
import pandas as pd
import os
```

```python
# Template_prompts
dataset_name_list = ["4chan", "Lexica", "Template", "coco"]
d_list = []
for dataset_name in dataset_name_list:
    data_file_path = f"unsafe_prompt_dataset/prompts/{dataset_name}_prompts.txt"
    with open(data_file_path, "r") as f:
        data = f.read()
    if "coco" in dataset_name:
        safe_label = 1
    else:
        safe_label = 0
    for line in data.split("\n"):
        d = {
            "dataset_name": dataset_name,
            "prompt": line,
            "safe_label": safe_label
        }
        d_list.append(d)
df = pd.DataFrame(d_list)
```

```python
df.to_pickle("unsafe_prompt_dataset/prompts.pkl")
```
